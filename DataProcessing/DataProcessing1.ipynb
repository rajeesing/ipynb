## Handling the missing data

```python
data = pd.read_csv('data.csv')
X = data.iloc[:,:-1].values
y = data.iloc[:,-1].values
print(data.isna().sum())
print(X)
```

```python
print(y)
```

### Taking Care of missing data

Handle missing data is one way to delete them if they are less in numbers like (1%) but another way is to fill the missing data with avg. value of that column.

```python
# sikitlearn library we are going to handle missing data
from sklearn.impute import SimpleImputer
imputer = SimpleImputer(missing_values=np.nan, strategy='mean')
imputer.fit(X[:,1:3])
X[:,1:3] = imputer.transform(X[:,1:3])  # transform method is used to replace missing data with the mean of the column
print(X)
```

### Encoding Categorical Data

#### Encoding the independent variable

We wanted to avoid the modal to have misinterpretation. Here Country is good candidate for categorical data.

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

column_trans = ColumnTransformer(
    [('encoder', OneHotEncoder(), [0])],    # the column numbers I want to apply this to
    remainder='passthrough'  # This leaves the rest of my columns in place
)

X = np.array(column_trans.fit_transform(X))
```

```python
print(X)
```

#### Encoding the dependent variable

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
y = le.fit_transform(y)
```

```python
print(y)
```

### Splitting the dataset into the Training Set and Test Set

Should we apply feature scaling before or after splitting the data? 

Ans. The answer is "After". They are two different data sets, because each set is exactly brand new model and test data should be a brand new data. If you apply before then you are going to get mean and SD then test is not any more fresh set of data.

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 1)
```

### Feature Scaling

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()

X_train[:, 3:] = sc.fit_transform(X_train[:, 3:]) # fit and transform on the training set
X_test[:, 3:] = sc.transform(X_test[:, 3:]) # only transform on the test set

print(X_train)
print(X_test)
```
