### Simple Linear Regression

This is simplest model where you have depedent variable (y) and independent variable (X1) with slope. b0 is intercept also known as constant and b1 slope coefficient. Here we are demonstrating the amount of potatos grow on the farm based on fertilizer we are going to use.

How do we know which slop line is best in our regression which can be determin by **Ordinary Least Squares**

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
data = pd.read_csv('../Data/Salary_Data.csv')
data.head()

x = data.iloc[:, :-1].values
y = data.iloc[:, 1].values
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=0)
```

```python
regression = LinearRegression()
regression.fit(x_train, y_train)
```

```python
plt.scatter(x_train, y_train, color='red')
plt.plot(x_train, regression.predict(x_train), color='blue')
plt.title('Salary vs Experience (Training set)')
```

```python
#using test data
plt.scatter(x_test, y_test, color='red')
plt.plot(x_train, regression.predict(x_train), color='blue')
plt.title('Salary vs Experience (Test set)')
```

#### How do I use my simple linear regression model to make a single prediction, for example, to predict the salary of an employee with 12 years of experience? 

```python
print(regression.predict([[12]]))
```

#### How do I get the final regression equation y = b0 + b1 x with the final values of the coefficients b0 and b1?

```python
print(regression.coef_)
print(regression.intercept_)
```
