### R Squared

Rule of thumb apply (highly dependent on the context) when R square is (should be between 0 and 1):

1.0     = Perfect fit (suspicious)

~0.9    = Very good

< 0.7   = Not great

< 0.4   = Terrible

< 0     = Model makes no sense for this data

### Adjusted R Squared

So when we add a new independent variable, few things will impact our regression. $SS_{tot}$ will not change and $SS_{res}$ will decrease or stay the same, because of Ordinary Least Squares.

### $Adj R^2 = 1 - (1-R^2) \times \frac{n - 1}{n - k - 1}$

k = number of independent variables, or number of columns

n = sample size, or number of rows
