```python
from pydub import AudioSegment
```

```python
def convert_opus_to_wav_pydub(opus_path, wav_path):
       """Converts an Opus file to WAV using pydub."""
       audio = AudioSegment.from_file(opus_path, format="ogg")
       audio.export(wav_path, format="wav")
```

```python
import os.path
```

```python
opus_file = '.\data\9eeab7ce-d4eb-4352-9f55-09f00c3cfa40.ogg'
if not os.path.exists(opus_file):
    print(f"File '{opus_file}' does not exist.")
else: 
    wav_file = '.\data\output.wav'
    convert_opus_to_wav_pydub(opus_file, wav_file)
```
