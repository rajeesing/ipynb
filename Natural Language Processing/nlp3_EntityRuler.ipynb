### What is Entity Ruler?

```python
import spacy
```

```python
nlp = spacy.load("en_core_web_sm")
text = "West Chestertenfieldville was referenced in Mr. Deeds."
doc = nlp(text) 
```

```python
print(doc)
```

```python
for ent in doc.ents:
    print(ent.text, ent.label_)
```

Above we are missing the whole context about `West Chestertenfieldville` which is basically not a real place but wanted to identify as place and label is currently assigned but considering if GPE would have not assigned correctly, then what?

```python
ruler = nlp.add_pipe("entity_ruler")
```

```python
nlp.analyze_pipes(pretty=True)
```

```python
patterns = [{"label": "GPE", "pattern": "West Chestertenfieldville"}] # GPE = Geopolitical Entity which is already in the model, but in case if it is not in the model, we can add it here.
```

```python
ruler.add_patterns(patterns)
```

```python
doc2 = nlp(text)
for ent in doc2.ents:
    print(ent.text, ent.label_) #nothing is changed because the entity ruler was identified after the ner pipeline was created and was already assigned
```

```python
nlp2 = spacy.load("en_core_web_sm")
ruler = nlp2.add_pipe("entity_ruler", before="ner")
ruler.add_patterns(patterns)    #this time the entity ruler is added before the ner pipeline is created so the entity is recognized
```

```python
doc = nlp2(text)
for ent in doc.ents:
    print(ent.text, ent.label_)
```

```python
# now lets change Deeds to Films not person
nlp3 = spacy.load("en_core_web_sm")
ruler = nlp3.add_pipe("entity_ruler", before="ner")
patterns = [{"label": "GPE", "pattern": "West Chestertenfieldville"}, {"label": "FILM", "pattern": "Mr. Deeds"}]
ruler.add_patterns(patterns)
```

```python
doc = nlp3(text)
for ent in doc.ents:
    print(ent.text, ent.label_) 
    
    # now Mr. Deeds is recognized as a film but may be some cases this is problem because in some cases Mr. Deeds can be a person
    # so we need to know when to consider as film and when to consider as person
```
