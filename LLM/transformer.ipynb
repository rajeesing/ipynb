There are different type of Neural Networks deal with different problems. 
**Convolutional Neural Networks**: It design to mimic human type brain which is good working with images like vision classification, analyzing images but often strugle with langage translation. 

Before Transformers we did have **Recurrent Neural Network** which takes one word sequence and process at a time. RNN holds the word in context as it waits for the next word. Often if you change the placement of the words, it may change whole meaning of the sentence. So it is important to maintain the order of the word. RNN really struggle when the input text is long. For example entire essay. It struggle understand the context in next sentence what was referred in previous sentence. For example in this sentence "The New York Times is a daily newspaper. **It** was first issued in 1851" where in second sentence **it** refers to `The New York Times` newspaper. In this case if we only able to look each token at time then we will not be able to truly understand the meaning of the text.

Transformers become more popular to solve this kind of problem where instead of word by word processing, it process entire sentence.

### Attention
Attention mechanism enables a model to weigh the importance of different words or tokens in an input sequence when producing an output. It give attention scores to each word in sentence and with higher attention score considered more important in model score.
